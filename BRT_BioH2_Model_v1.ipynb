```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
# Parameters
k_bio = 1.2          # mmol/m²/h
P_c = 0.30           # critical porosity
beta = 1.5           # scaling exponent
alpha = 0.05         # salinity inhibition
```

```python
# Porosity (0.2 to 0.6)
P = np.linspace(0.2, 0.6, 100)

# Salinity (0 to 35 ppt)
S = np.linspace(0, 35, 100)

P_grid, S_grid = np.meshgrid(P, S)
```

```python
# Initialize flux
J = np.zeros_like(P_grid)

# Apply model
mask = P_grid >= P_c
J[mask] = k_bio * ((P_grid[mask] - P_c) ** beta) * np.exp(-alpha * S_grid[mask])
```

```python
plt.figure(figsize=(6,5))

plt.imshow(J, extent=[0.2, 0.6, 0, 35],
           origin='lower', aspect='auto',
           cmap='cividis')

plt.xlabel('Porosity')
plt.ylabel('Salinity (ppt)')
plt.title('Regime Map of Hydrogen Production')

cbar = plt.colorbar()
cbar.set_label('H2 Flux (mmol/m²/h)')

plt.tight_layout()

plt.savefig("figure1_heatmap.svg", format="svg", dpi=300)

plt.show()
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

plt.rcParams.update({
    "font.size": 12,
    "axes.labelsize": 12,
    "axes.titlesize": 14
})

fig = plt.figure(figsize=(7,5))
ax = fig.add_subplot(111, projection='3d')

# Surface plot with CIVIDIS
surf = ax.plot_surface(P_grid, S_grid, J, cmap='cividis')

# Labels
ax.set_xlabel('Porosity')
ax.set_ylabel('Salinity (ppt)')
ax.set_zlabel('H2 Flux (mmol/m²/h)')
ax.set_title('Biohydrogen Flux in Mangrove Sediments')

# Color bar (VERY IMPORTANT)
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10, label='H2 Flux (mmol/m²/h)')
surf = ax.plot_surface(P_grid, S_grid, J, cmap='cividis', linewidth=0, antialiased=True)
plt.tight_layout()

# Save files
plt.savefig("figure2_surface.svg", format="svg", dpi=300)
plt.savefig('/content/drive/MyDrive/figure2_heatmap.svg', format='svg', dpi=300)

plt.show()
```
